```python
#import dataset
import pandas  as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
#Load dataset
data = pd.read_csv("student_marks.csv")
```

```python
#Input and output
x = data[['Hours']]
y = data[['Marks']]
```

```python
#Split dataset
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=1
)
```

```python
#Train model
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
#Display accuracy
print("Training Accuracy:",
      round(model.score(x_train, y_train) * 100, 2))
print("Teasting Accuracy:",
      round(model.score(x_test, y_test) * 100, 2))
```

```python
#predict marks for 11 study hours
new_data = pd.DataFrame([[11]], columns=['Hours'])
pred = model.predict(new_data)
print("Predict Marks for 11 hours:", round(pred[0][0], 2))
```

#Q.2 A company has collected the following employee data:
• Employee ID
• Employee Age
• Years of Experience
• Salary
You need to develop a Linear Regression model to predict an employee's salary.

```python
#Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
#Load the dataset
data = pd.read_csv("employee_salary.csv")
```

```python
#Independent (X) and Dependent (Y) variables
X = data[['Years_of_Experience']]
Y = data['Salary']
```

```python
#Split the dataset into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=1
)
```

```python
#Create the Linear Regression model
model = LinearRegression()

#Train the model
model.fit(X_train, Y_train)
```

```python
#Predict salary for the test data
Y_pred = model.predict(X_test)
```

```python
#Predict salary for an employee with 7 years of experience
salary = model.predict([[7]])
print("\nPredicted Salary for 7 Years of Experience:", salary[0])
```

```python
#Display Slope, Intercept and R² Score
print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
print("R² Score:", r2_score(Y_test, Y_pred))
```

```python
#Plot the regression line
plt.scatter(X, Y, color='blue', label='Actual Data')
plt.plot(X, model.predict(X), color='red', linewidth=2, label='Regression Line')

plt.title("Employee Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

# FOR California Housing Dataset #

```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
#Load dataset
data = pd.read_csv("california_housing.csv")
```

```python
#Check for missing values
print(data.isnull().sum())
```

```python
#Select Independent (X) and Dependent (Y) variables
X = data[['MedInc']]
Y = data['MedHouseVal']
```

```python
#Split the dataset
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=1
)
```

```python
#Create and train the model
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
#Predict house prices
Y_pred = model.predict(X_test)
```

```python
#Display Coefficient, Intercept and R² Score
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R² Score:", r2_score(Y_test, Y_pred))
```

```python
#Plot the regression line
plt.scatter(X, Y, color="blue")
plt.plot(X, model.predict(X), color="red")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.title("Linear Regression")
plt.show()
```

```python
# Interpret the results.
print("Higher Median Income generally leads to a higher Median House Value.")
```

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
#load MNIST-like digits dataset
digits = load_digits()
```

```python
x = digits.data
y = digits.target
```

```python
#Binary Classification
#true if digit is 5,otherwise False
y = (y==5)
```

```python
#Split dataset
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=1
)
```

```python
#Train Model
model = LogisticRegression(max_iter=10000)
model.fit(x_train, y_train)
```

```python
#Accuracy 
print("Training Accuracy:", model.score(x_train, y_train)*100)
print("Testing Accuracy:", model.score(x_test, y_test)*100)
```

```python
#predict first test sample
prediction = model.predict([x_test[0]])

if prediction[0]:
    print("The digit is 5.")
else:
    print("The digit is not 5.")
```
